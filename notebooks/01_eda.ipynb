```python
df['no_show'] = df.no_show.apply(lambda x: 1 if x == "yes" else 0)
df.head(5)
# df.no_show.dropna(inplace=True)
```

```python
counts = df.groupby(['no_show', 'appointment_shift']).size().unstack()
counts
```

```python
counts = df.groupby(['appointment_time', 'no_show']).size().unstack().sort_values(by=0, ascending=False).head(20)   
counts
```

```python
# Calculate the correlation matrix with the numerical columns
df_numerical = df.select_dtypes(include=[np.number])
correlation_matrix = df_numerical.corr()
# Plot the correlation matrix
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

```python
df.icd.isnull().sum()
```

```python
df.age.describe()
plt.figure(figsize=(12, 5))

# 年齡直方圖
plt.subplot(1, 2, 1)
sns.histplot(data=df, x='age', bins=30)
plt.title('Age Distribution')

# 年齡箱型圖（按no-show分組）
plt.subplot(1, 2, 2)
sns.boxplot(x='no_show', y='age', data=df)
plt.title('Age by No-show Status')
plt.show()

# 3. 看不同年齡段的no-show率
temp_bins = [0, 18, 30, 45, 60, 75, 90, 100]
df['temp_age_group'] = pd.cut(df['age'], bins=temp_bins)
no_show_by_age = df.groupby('temp_age_group')['no_show'].mean()
print("\nNo-show rate by age group:")
print(no_show_by_age)
```
