## Data Cleaning 

```python
df.isnull().sum()
```

```python
### Handle Missing Values:
## Specialty: filled with "unknown or mode"
## No-show Reasons: missing most of the values -> drop this column
df = df.drop(columns=['no_show_reason'])
## City: filled with unknown
df.city = df.city.fillna('unknown')
df.city.value_counts()
```

```python
df.disability.fillna("none", inplace=True)
```

```python
df = df.drop(columns=['icd'])
```

```python
df.specialty = df.specialty.fillna('unknown')
df.specialty.value_counts()
```

```python
# Check duplicates
df[df.duplicated()]
```

```python
## Remove the duplicates
df.drop_duplicates(inplace=True)
```

```python
## Drop age and date of birth
df.drop(columns=['age', 'date_of_birth'], inplace=True)
```

```python
df = df.drop(columns=['entry_service_date'])
```

```python
# fill missing average weather data with the mean temperature of that city from other days
weather_cols = ['average_temp_day', 'average_rain_day', 'max_temp_day', 'max_rain_day']

for col in weather_cols: 
    df[col] = df.groupby(['city', 'appointment_month'])[col].transform(lambda x: x.fillna(x.mean()))
```

###  Feature Engineering

```python

# Save the cleaned dataset
df.to_csv('../data/cleaned_data.csv', index=False)
```

```python
cleaned_df = pd.read_csv('../data/cleaned_data.csv')
cleaned_df.head(5)
```

```python
cleaned_df.nunique()
# Save the cleaned dataset
```

```python
cleaned_df.rain_intensity.value_counts()
```

```python

cleaned_df.gender = cleaned_df.gender.map({'M': 0, 'F': 1})
```

```python
cleaned_df.no_show = cleaned_df.no_show.map({'yes': 1, 'no': 0})
```

```python
cleaned_df.info()
```

```python
df.rain_intensity.unique()
```

```python
rain_intensity = {'no_rain': 0, 'weak':1, 'moderate': 2, 'heavy': 3}
cleaned_df.rain_intensity = cleaned_df.rain_intensity.map(rain_intensity)
```

```python
df.heat_intensity.unique()
```

```python
# 'mild', 'cold', 'warm', 'heavy_cold', 'heavy_warm'
heat_intensity = {'heavy_cold': 0, 'cold': 1, 'mild': 2, 'warm': 3, 'heavy_warm': 4}
cleaned_df.heat_intensity = cleaned_df.heat_intensity.map(heat_intensity)
cleaned_df.head(5)
```

```python
cleaned_df.appointment_shift.unique()
# 'morning', 'afternoon'
cleaned_df.appointment_shift = cleaned_df.appointment_shift.map({'morning': 0, 'afternoon': 1})
```

```python
cleaned_df['appointment_date'] = pd.to_datetime(cleaned_df['appointment_date'], format='%d/%m/%Y')
cleaned_df['appointment_time'] = pd.to_datetime(cleaned_df['appointment_time'], format='%H:%M').dt.time
cleaned_df.head(5)
```

```python
# Extract Month and Day from the appointment date
cleaned_df['day_of_week'] = cleaned_df['appointment_date'].dt.dayofweek
cleaned_df['is_weekend'] = cleaned_df['day_of_week'].isin([5, 6]).astype(int)
cleaned_df['appointment_month'] = cleaned_df['appointment_date'].dt.month
cleaned_df['appointment_year'] = cleaned_df['appointment_date'].dt.year
cleaned_df['appointment_day'] = cleaned_df['appointment_date'].dt.day

```

```python
cleaned_df
```

```python
cleaned_df['appointment_hour'] = pd.to_datetime(cleaned_df['appointment_time'], format='%H:%M:%S').dt.hour
```

```python
cleaned_df
```

```python
cleaned_df.info()
```

```python
cleaned_df = pd.get_dummies(cleaned_df, columns=['city', 'specialty', 'disability'], drop_first=True)
```

```python
cleaned_df.info()
```

```python
cleaned_df.drop(columns=['appointment_date', 'appointment_time'], inplace=True)
```

```python
cleaned_df.drop(columns=['no_show_num'], inplace=True)
```

```python
cleaned_df.info()
```

```python
cleaned_df.nunique()
cleaned_df.gender.unique()
# Drop the rows with unknown gender
cleaned_df = cleaned_df.dropna(subset=['gender'])
```

```python
# Save the cleaned and Encoded dataset
cleaned_df.to_csv('../data/cleaned_encoded_data.csv', index=False)
```
